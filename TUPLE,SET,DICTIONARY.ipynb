SET OPERATION

UNION

```python
A = {1,2,3,4,5}
B = {4,5,6,7,8}
C = {8,9,10}
```

```python
A | B # Union of A and B (All element from both sets. No Duplicates)
```

```python
A.union(B) # Union of A and B
```

```python
A.union(B, C) # Union of A, B and C.
```

```python
"""
Updates the set calling the update() method union of A, B & C.

For below example Set A will be updated with union of A,B & C.
"""
A.update(B,C)
A
```

Intersection

```python
A = {1,2,3,4,5}
B = {4,5,6,7,8}
```

```python
A & B # Intersection of A and B (Common item in both sets)
```

```python
A.intersection(B) Intersection of A and B
```

```python
"""
Updates the set Calling the intersection_update() method with the intersection

for below example Set A will be Updated with the intersection of A & B.
"""
A.intersection_update(B)
A
```

Differnce

```python
A = {1,2,3,4,5}
B = {4,5,6,7,8}
```

```python
A - B # Set of element that are only in A but not in B
```

```python
A.difference(B) # D iffernce of sets
```

```python
B - A # set of the element that are only in B but not in A
```

```python
A.difference(B) # Differnce od sets
```

```python
B.difference(A)
```

```python
"""
Update the set calling the difference_update() metthod with the difference of set 

for below example Set B will be updated with the difference of B & A.
"""
B.difference_update(A)
B
```

Symmetric Difference

```python
A = {1,2,3,4,5}
B = {4,5,6,7,8}
```

```python
A ^ B # Symmetric difference (Set of element in A and B but not in both> 
```
