All / Any

# True - if all element in a list are true
# False - if any element in a list is false

```python
l1 = [1,2,3,4,0]
```

```python
all(l1) # will return false as one value id false (value 0)
```

```python
any(l1) # will return true as we have items in the list with true value 
```
